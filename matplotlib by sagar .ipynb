```python
#plt.bar(x,height, color'"color name",width= value, label="label name")
product=['A','B','C','D','E']
sales=[1200,100,1500,800,1000]
plt.barh(product,sales, color ="orange", label="Sales 2025")
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title("Product Sales comparsion")
plt.legend()
plt.show()
```

```python
#plt.pie(x, label="label_list",color="color_list",autopct='%1.1f%%)

region=['N','S','E','W']
revenue=[3000,2000,1500,1000]
plt.pie(revenue, colors=["orange","gold","skyblue","green"], labels=region, autopct='%1.1f%%')

plt.title("REVENUE CONTRIBUTATION BY REGION")

plt.show()
```

```python
#plt.hist(data,bins=number+of_bins, color="colorname',edgecolor="black")
scores=[45,56,78,97,47,5,30,84,93,68]
plt.hist(scores,bins=5, color="purple",edgecolor="black")

plt.xlabel('SCORES RANGE')
plt.ylabel('NUMBER OF STUDENTS')
plt.title('SCORE DISTRUBATION OF STUDENTS')
plt.show()
```

```python
#plt.scatter(x,y,color="color_name", marker="marker",label="label name")
students=[1,2,3,4,5]
examscores=[50,69,69,76,87]
plt.scatter(students,examscores,color="green", marker="o",label="students data")
plt.xlabel("HOURS STUDY")
plt.ylabel("Exam score")
plt.title("relationship between student and exam")
plt.legend()
plt.grid()
plt.show()
```

```python
#subplot

x=[1,2,3,4]
y=[10,25,20,15]

plt.subplot(1,2,1)# row,col, position of subplot
plt.plot(x,y)
plt.title("LINE CHART")

plt.subplot(1,2,2)# row,col, position of subplot
plt.bar(x,y)
plt.title("BAR CHART")
plt.show()
```

```python
x=[1,2,3,4]
y=[10,25,20,15]

#fig,ax =plt.subplot(nrows,ncols, figsize=(width,height))
fig,ax =plt.subplots(1,2, figsize=(10,5))

ax[0].plot(x,y)
ax[0].set_title('Line Plot')

ax[1].bar(x,y)
ax[1].set_title('Bar chart')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
data = pd.read_csv("country_wise_latest.csv")

# Display basic info
print(data.head())
print(data.describe())

# Data Cleaning
data = data.dropna()

# Visualization
plt.figure(figsize=(12,6))
plt.plot(data["Country/Region"], data["New cases"], label="New Cases", color='blue')
plt.title("COVID-19 New Cases by Country")
plt.xlabel("Country/Region")
plt.ylabel("New Cases")
plt.xticks(rotation=90)  # Helps read country names on x-axis
plt.legend()
plt.tight_layout()       # Adjusts layout to prevent overlap
plt.show()
```
