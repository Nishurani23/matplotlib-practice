```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.linspace(0,5,11)
y=x**2
```

```python
plt.plot(x,y)
plt.title('abc')

plt.show()
```

```python
plt.subplot(2,2,1)
plt.subplot(2,2,2)

plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x,y)
```

```python
fig=plt.figure()
axis1=fig.add_axes([0.1,0.1,3,3])
axis1.plot(x,y)
axis2=fig.add_axes([0.3,0.8,1,1])
axis2.plot(x,y)
```
