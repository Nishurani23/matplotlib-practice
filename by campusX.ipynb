```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#plotting a single function
price=[48000,54000,57000,49000,47000,45000]
year=[2015,2016,2017,2018,2019,2020]
plt.plot(year,price)
```

```python
#from a pandas dataframes
batsman=pd.read_csv('sharma-kohli.csv')
batsman
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
#labels, title
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title("Compare Vkholi vs RG sharma")
plt.xlabel('Season')
plt.ylabel('Runs scores')
```

```python
#color and line(width and style) and marker(size)
#legend
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle= 'dashed',marker='o' ,label='virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='blue',linestyle= 'dashdot',marker='d',label='rohit')         
plt.title("Compare Vkholi vs RG sharma")
plt.xlabel('Season')
plt.ylabel('Runs scores')
plt.plot(batsman['index'],batsman['RG Sharma'],color='blue')
plt.title("Compare Vkholi vs RG sharma")
plt.xlabel('Season')
plt.ylabel('Runs scores')
plt.legend()
```

```python
#limiting axes
price=[48000,54000,57000,49000,47000,45000000]
year=[2015,2016,2017,2018,2019,2020]
plt.plot(year,price)
plt.ylim(0,75000)
plt.xlim(2017,2019)
```

```python
#scatter  plots
#use cases- Finding correlations
#numerical vs numerical

#plt.scatter simple function
x=np.linspace(-10,10,50)
y=10*x+3+np.random.randint(0,300,50)
plt.scatter(x,y)
```

```python
# plt.scatter on datasets
df=pd.read_csv('batter.csv')
df=df.head(50)
df
```

```python
plt.scatter(df['avg'],df['strike_rate'],color='red')
plt.title("Avg and strike rate analysis of top 50 batsman")
plt.xlabel("Average")
plt.ylabel("SR")
```

```python
#size
#slow
tips=sns.load_dataset('tips')
plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*20)
```

```python
#scatterplot using plt.plot
#faster
plt.plot(tips['total_bill'],tips['tip'],'o')
```

```python
#bar chart --- numerical vs catergorical----- use cases- aggregate analysis of groups

#simple chart
children=[10,20,34,65,89]
colors=['red','blue','green','yellow','pink']
plt.bar(colors,children,color=['red','blue','green','yellow','pink'])

#horizontal bar chart
#plt.barh(colors,children,color=['red','blue','green','yellow','pink'])
```

```python
# bar using datasets
dff=pd.read_csv("batsman_season_record.csv")
dff
```

```python
#color or label
plt.bar(dff['batsman'],dff['2015'] ,label='2015')
plt.bar(dff['batsman'],dff['2016'],label='2016')
plt.bar(dff['batsman'],dff['2017'],label='2017')

plt.legend()
```

```python
dataa=dff.groupby('batsman')[['2015','2016','2017']].sum()
dataa.plot(
    kind='bar',
    figsize=(12,6)
)

plt.title('batsman Analysis')
plt.xlabel('players')
plt.ylabel('Runs')
plt.xticks(rotation=45)
plt.show()
```

```python
#pie chart----categorical vs numerical--use cases--to find contibution on a standard scale

#simple data
data=[23,45,100,20,49]
subjects=['eng','hindi','sci','sst','maths']
plt.pie(data,labels=subjects)
plt.show()
```

```python
df=pd.read_csv('gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%1.1f%%')
plt.show
```

```python
#percentage and color
#explode shadow
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%1.1f%%',explode=[0.1,0,0,0,0,0])
plt.show
```
