```python
import matplotlib.pyplot as plt
```

```python
#create first chart
x=[1,2,3,4,5]
y=[10,20,30,40,50]
plt.plot(x,y)
plt.show()
```

```python
# customize chart
plt.plot(x,y,color='green',marker='v',linestyle='dashed',linewidth=2,markersize=12)
plt.show()
```

```python
plt.figure(figsize=(4,3))
plt.plot(x,y,color='red',marker='.',linestyle='dashed',linewidth=2,markersize=12)
plt.title('LINE CHART')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.show()
```

```python
# Adavanced line chart
plt.figure(figsize=(4,3))

x=[1,2,3,4,5]
y1=[10,20,30,40,50]
y2=[15,20,35,40,55]

plt.plot(x,y1, label='sales 2024')
plt.plot(x,y2, label='sales 2025')
plt.title('YOY SALES')
plt.xlabel("months")
plt.ylabel("sales")

plt.legend()
plt.show()
```

```python
#bar chart
plt.figure(figsize=(4,3))
x=[1,2,3,4,5]
y=[15,20,35,40,55]
plt.bar(x,y)
plt.title("BAR GRAPH")
plt.show()
```

```python
#histogram
plt.figure(figsize=(4,3))
import random
data=[random.randint(1,10) for _ in range(100)]
plt.hist(data,bins=4)
plt.title("histogram graph")
plt.show()
```

```python
#pie chart
categories=['A','B','C','D','E']
sales=[10,20,30,40,50]
plt.pie(sales, labels=categories, autopct= '%1.1f%%')
plt.title("pie chart")
plt.show()
```

```python
#scatter plot chart
plt.figure(figsize=(4,3))

y1=[10,20,30,40,50]
y2=[15,20,35,40,55]
plt.scatter(y1,y2)
plt.title("scatter")
plt.show()
```

```python
#subplot
plt.figure(figsize=(4,3))

categories=['A','B','C','D','E']
sales=[10,20,30,40,50]

y1=[10,20,30,40,50]
y2=[15,20,35,40,55]
#first chart-bar chart
plt.subplot(1,2,1)#row,col,position
plt.bar(categories,sales)
plt.title("daily wages")
plt.xlabel("Week Days")
plt.ylabel("sales")

#second chart-- scatter plot
plt.subplot(1,2,2)#row,col,position
plt.scatter(y1,y2)
plt.title("user example")
plt.xlabel("Week Days")
plt.ylabel("sales")
plt.show()
```

```python
#matplotlib with pandas
#create df
import pandas as pd
data={
    'Month' : ['jan','feb','mar','april'],
    'sales': [1200,1100,1300,2400]
}
df=pd.DataFrame(data)
df
```

```python
plt.figure(figsize=(4,3))
plt.bar(df['Month'],df['sales'])
plt.title("matplotlib with pandas")
plt.xlabel("Month")
plt.ylabel("sales")
plt.show()
```

```python
plt.figure(figsize=(4,3))
plt.bar(df['Month'],df['sales'])
plt.title("matplotlib with pandas")
plt.xlabel("Month")
plt.ylabel("sales")

plt.savefig("Monthly_sales_from_df.png")

plt.show()
```
